```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_excel("imdb_ratings.xlsx")
df.head()
```

```python
df.info()
```

```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
le = LabelEncoder()
df['titleId_encoded'] = le.fit_transform(df['titleId'])
```

```python
scaler = StandardScaler()

df[['averageRating', 'numVotes']] = scaler.fit_transform(
    df[['averageRating', 'numVotes']]
)
```

```python
df['vote_category'] = pd.cut(
    df['numVotes'],
    bins=[-2, -0.5, 0, 0.5, 3],
    labels=['Very Low', 'Low', 'Medium', 'High']
)
```

```python
df['rating_class'] = pd.cut(
    df['averageRating'],
    bins=[-2, -0.5, 0.5, 3],
    labels=['Low', 'Average', 'High']
)
```

```python
df = pd.get_dummies(
    df,
    columns=['vote_category', 'rating_class'],
    drop_first=True
)
```

```python
df.corr(numeric_only=True)
```

```python
df.corr(numeric_only=True)['averageRating'].sort_values(ascending=False)
```

```python
df_model = df.drop(columns=['title', 'titleId'])
df_model.head()
```

```python
df_model.shape
```

```python
df_model.to_csv("imdb_task3_transformed.csv", index=False)
```

In Task 3, feature understanding and data transformation were performed on the IMDb ratings dataset. Irrelevant columns were removed, categorical variables were encoded using Label Encoding and One-Hot Encoding, and numerical features were standardized using StandardScaler. Additional features such as vote categories and rating classes were engineered to enhance interpretability. Correlation analysis was conducted to identify important features. The final transformed dataset is clean, normalized, and ready for analysis and machine learning modeling.
